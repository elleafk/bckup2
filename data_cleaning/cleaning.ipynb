```python
df["First Name"] = df["First Name"].str.title()
df["First Name"].unique()
```

```python
df["Last Name"] = df["Last Name"].str.title()
df["Last Name"].unique()
```

```python
def cleanGrade(value):
    value = str(value).lower()
    
    if "9" in value:
        return "Grade 9"
    elif "10" in value:
        return "Grade 10"
    elif "11" in value:
        return "Grade 11"
    elif "12" in value:
        return "Grade 12"
    return value

df["Grade Level"] = df["Grade Level"].apply(cleanGrade)
df["Grade Level"].unique()
```

```python
df.dtypes
```

```python
df.isnull().sum()

clean_df = df.dropna(
    subset = ["Math Grade",
        "Science Grade",
        "English Grade",
        "Attendance Days",
        "Total School Days"
    ]
)

clean_df.isnull().sum()
```

```python
clean_df["Enrollment Status"] = clean_df["Enrollment Status"].str.capitalize()
clean_df["Enrollment Status"].unique()
```

```python
clean_df[clean_df.duplicated(subset = ["Student ID"], keep=False)]
```

```python
clean_df["Contact Phone"] = clean_df["Contact Phone"].str.replace(r"\D","", regex=True)
```

```python
clean_df[clean_df.duplicated(keep=False)]
clean_df = clean_df.drop_duplicates()
```
