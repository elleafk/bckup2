```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_excel("CLEANED.xlsx")
```

```python
display(df)
```

```python
plt.boxplot(df["Attendance"], patch_artist=True, boxprops=dict(facecolor="skyblue"), medianprops=dict(color="red"))

plt.title("Attendance Box Plot")
plt.ylabel("Attendance Percentage")

plt.show()
```
