```python
df["Program"] = df["Program"].replace({
    "Bachelor of Science in IT": "BSIT"
})

df["Program"].unique()
```

```python
df[df.duplicated(subset=["Student_ID"], keep=False)]
```

```python
df = df.drop_duplicates(subset=["Student_ID"])

display(df)
```

```python
df = df.dropna(subset=["GPA"])

display(df)
```

```python
df["Attendance"] = df["Attendance"].clip(0,100)
display(df)
```

```python
programStudents = df["Program"].value_counts()

plt.bar(programStudents.index, programStudents.values, color="maroon")

plt.title("Number of Students by Program")
plt.xlabel("Students")
plt.ylabel("Programs")

plt.show()
```
