```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_excel("CLEANED.xlsx")

display(df)
```

```python
plt.scatter(df["Attendance"],df["GPA"], marker="o", color="purple", s=200, edgecolors="black")

plt.title("Relationship Between Attendance and GPA")
plt.xlabel("Attendance Percentage")
plt.ylabel("GPA")
plt.grid(True)
plt.show()
```
